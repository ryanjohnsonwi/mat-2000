```python
import requests
import re
from bs4 import BeautifulSoup, Tag
import pandas as pd
import pickle
import html2text
h = html2text.HTML2Text()
```

```python
# Opening the html file
HTMLFile = open("Section 2.2 - MAT 2000_ Introduction to Statistics _ zyBooks.html", "r", encoding='utf-8')
  
# Reading the file
index = HTMLFile.read()
  
# Creating a BeautifulSoup object and specifying the parser
S = BeautifulSoup(index)

HTMLFile.close()
```

```python
#print(S.prettify)
```

```python
tags = S.select('title, h1, h3, p, table, ul, ol, img, div.activity-title, div.static-container-title')
#tags
```

```python
t = tags[0]
print(t)
print(['bread-crumbs'] in t.attrs.values())
```

```python
#Title
title = tags.pop(0).get_text().replace(' | zyBooks', '\n')

#Subtitle
subtitle = '## Grace College using Zybooks\n'

#Create the top of the Markdown document
top = '# ' + title + subtitle

print(top)
```

```python
#filename
filename = title
filename = re.sub(r'(S)ection (\d+)\.(\d+)(.+)', r'\1ection_\2_\3', filename).lower()[:-1] + ".md"

print(filename)
```

```python
#Turn list of tags into long string of html tags
new_html = ''
for t in tags:
    if t.name == 'div':
        t.name = 'h3' #Change name to h3
        att = list(t.attrs)  #list the attributes
        for a in att:   #so that we can
            del t[a]    #delete them
        t.string = re.sub("\s+", " ", t.text) #trim extra white space
    elif "Keep your answer in mind as you continue" in t.text:
        continue
    elif 'zyBooks' in t.attrs.values():
        continue #Do not include the zyBooks logo
    elif ['bread-crumbs'] in t.attrs.values():
        continue #Do not include the breadcrumbs
    elif ['level-1'] in t.attrs.values():
        continue #Do not include the level-1 sidebar
    elif ['level-2'] in t.attrs.values():
        continue #Do not include the level-2 sidebar
    s = str(t)
    new_html += s
```

```python
new_html[:800]
```

```python
text = top + h.handle(new_html)
```

```python
md = open(filename, 'w', encoding='utf-8')
md.write(text)
md.close()
```
